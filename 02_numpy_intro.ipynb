## 1.2. Matrix Operations

```python
np.zeros((3,5))
```

```python
np.ones((5, 3))
```

```python
a = np.arange(1, 7)
a
```

```python
a.shape
```

```python
a[3] = 7
a
```

```python
a[:5] = 1
a
```

```python
a[1:5] = [5, 9, 6, 8]
a
```

```python

# This is most useful
b = np.zeros((2, 2))
b[0, 0] = 1
b[0, 1] = 2
b[1, 0] = 3
b[1, 1] = 4

b
```

```python
b.shape
```
