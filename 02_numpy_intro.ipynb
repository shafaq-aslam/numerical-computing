# 2. Misc Operations with Numpy

```python
import math

def basic_sigmoid(x):
    """
    Compute sigmoid of x.

    Arguments: 
    x -- A scalar

    Return:
    s -- sigmoid(x)
    """

    s = 1./(1. + math.e ** (-x))
    return s
```

```python
basic_sigmoid(-1)
```

```python
basic_sigmoid(0)
```

```python
basic_sigmoid(1)
```

```python
basic_sigmoid(8)
```

```python
x = [-1, 0, 3]
basic_sigmoid(x) # why? Because this function is designed for scalars, not vector
```

```python
import numpy as np

x = [-1, 0, 3]
x = np.array(x)
basic_sigmoid(x)
```

## 2.1. Broadcasting

```python
x = np.array([1, 2, 3])
x * 3
```

```python
x + 3
```

*General Rules of Broadcasting:*</br>
    <p>Two dimensions are compatible, when</p>
    <ul>
        <li>They are equal, or</li>
        <li>one of them is 1</li>
    </ul>

```python
x = np.arange(5)
x_r = x.reshape(5, 1)
y = np.ones(6)
z = np.ones((3, 4))

print("x = ",x)
print("x reshape = ",x_r)
print("y = ",y)
print("z = ",z)

print("Shapes: ")
print(x.shape)
print(x_r.shape)
print(y.shape)
print(z.shape)
```

```python
x + y # it will not work, because the sizes are not equal, so it can't be broadcast
```

```python
print("x_r = ",x_r)
print("y = ",y)
```

```python
x_r + y # x_r is broadcast over the columns of y
```

```python
y + x_r
```

```python
np.array([1]) + y
```

```python
print(z)
z.shape
```

```python
x
```

```python
z + x # none of rule is satisfied
```

```python
p = np.array([0, 1, 2, 4])
p.shape
```

```python
z + p # first rule of broadcast is satisfied here
```
