## 1.2. Vectors

A <strong>vector</strong> in R<sup>n</sup> is an <i>n</i>-tuple. They can be horizontal and are called row vectors or vertical in which case they are called column vectors (the default).

## 1.3. Vector Norm

The norm of a vector is a measure of its length. There are many ways of defining the length of a vector depending on the metric used (i.e., the distance formula chosen). So, let's take a look at some distance formulae first.

### 1.3.1. Euclidean Distance

Euclidean distance is what you are already familiar with. The Euclidean distance between two points 
$$(x_1,y_1) $$ and $$(x_2,y_2)$$ is given as:

The Euclidean distance between two points is:

$$d = \sqrt{(x_2 - x_1)^2 + (y_2 - y_1)^2}$$

For 3D points:

$$d = \sqrt{(x_2 - x_1)^2 + (y_2 - y_1)^2 + (z_2 - z_1)^2}$$

If we take any vector and take it's x- and y-coordinates as its components, we can redefine the above like so:

This length is then called the L-2 Norm of the vector -- the stuff you know as the magnitude. The specific formula then is:

$$\|v\|_2 = \sqrt{\sum_{i} v_i^2}$$
where 
$$v_i$$
 are the elements of the vector 
v
.

## 1.4. Numpy Vectors Revisited

```python
import numpy as np
from numpy.linalg import norm
```

```python
# Row Vector
vector_row = np.array([[1, -5, 3, 2, 4]])
print(f'Row Vector: {vector_row}\n')

# Transpose of the Row vector
new_vector = vector_row.T
print(f'Transpose of Row Vector: {new_vector}\n')

norm_1 = norm(new_vector, 1)
norm_2 = norm(new_vector, 2)
norm_inf = norm(new_vector, np.inf) # norm infinity

print('L_1 is: %.1f'%norm_1)
print('L_2 is: %.1f'%norm_2)
```

### 1.4.1. Vector Operations

```python
u = np.array([[0, 3, 2]])
v = np.array([[4, 1, 1]])
w = np.array([[0, -2, 0]])

x = 3*u - 2*v + 4*w
print(x)
```

```python
print(u,'\n', v)
```

```python
print(v.T)
```

```python
u.dot(v) # bcz shapes are not aligned
```

```python
print(u.shape)
print(v.shape)
```

```python
u = np.array([[0, 3, 2]])
v = np.array([4, 1, 1]) # this is rank 1 tensor

x = u * v
print(x)
```

```python
print(u, '\n', v)
print(v.T)

u.dot(v)
```

```python
print(u.shape)
print(v.shape)
```

## 1.5. Matrices

### 1.5.1 Matrix Multiplication

```python
P = np.array([[1, 7], [2, 3], [5, 0]])
Q = np.array([[2, 6, 3, 1], [1, 2, 3, 4]])
print(P)
print(Q)
print(np.dot(P, Q))
```
