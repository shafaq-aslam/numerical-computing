# 2. Misc Operations with Numpy

```python
import math

def basic_sigmoid(x):
    """
    Compute sigmoid of x.

    Arguments: 
    x -- A scalar

    Return:
    s -- sigmoid(x)
    """

    s = 1./(1. + math.e ** (-x))
    return s
```

```python
basic_sigmoid(-1)
```

```python
basic_sigmoid(0)
```

```python
basic_sigmoid(1)
```

```python
basic_sigmoid(8)
```

```python
x = [-1, 0, 3]
basic_sigmoid(x) # why? Because this function is designed for scalars, not vector
```

```python
import numpy as np

x = [-1, 0, 3]
x = np.array(x)
basic_sigmoid(x)
```

*General Rules of Broadcasting:*</br>
    <p>Two dimensions are compatible, when</p>
    <ul>
        <li>They are equal, or</li>
        <li>one of them is 1</li>
    </ul>

```python
x = np.arange(5)
x_r = x.reshape(5, 1)
y = np.ones(6)
z = np.ones((3, 4))

print("x = ",x)
print("x reshape = ",x_r)
print("y = ",y)
print("z = ",z)

print("Shapes: ")
print(x.shape)
print(x_r.shape)
print(y.shape)
print(z.shape)
```

```python
x + y # it will not work, because the sizes are not equal, so it can't be broadcast
```

## 2.2. Normalizing Rows and Columns

Often, we wish to normalize value of a row to fall within a specific range.

```python
x = np.array([[0, 3, 4],
             [1, 6, 4]])
```

```python
# There are many ways, but one easy way to normalize is that divide the elements with max
x_norm = np.amax(x, axis=1) # to get row-wise max
print(x_norm) 
```

```python
x / x_norm  # it couldn't be broadcast
```

```python
print(x.shape)
print(x_norm.shape)
```

```python
x_norm = x_norm.reshape(2, 1)
```

```python
x / x_norm
```

```python
# We can also use another normalization method
x = np.array([[0, 3, 4],
             [1, 6, 4]])

# no need to reshape
x_norm = np.linalg.norm(x, ord = 2, axis=1, keepdims= True)
x / x_norm
```

## 2.4. Vectorization

```python
dim = 10000 # try increasing this after running experimet
```

```python
A = np.random.rand(dim, dim)
B = np.random.rand(dim, dim)
```

```python
A[0].size, A[1].size
```

```python
def add_arrays(A, B):
    C = np.zeros((A[0].size, A[1].size))

    for i in range(A[0].size):
        for j in range(A[1].size):
            C[i, j] = A[i, j] + B[i, j]
    return C
```

```python
%time C = add_arrays(A, B)
```

```python
%time C = A + B
```

```python
import time

#Non-vectorized time
start_time = time.time()

C = add_arrays(A, B)

end_time = time.time()
non_vec_time = end_time - start_time

#Vectorized Time
start_time = time.time()

C = A + B

end_time = time.time()
vec_time = end_time - start_time
```

```python
vec_time / non_vec_time * 100
```
