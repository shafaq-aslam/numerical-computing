# 1. Vector/Matrix Definition & Usage

## 1.1 Vector Representation

```python
class Vector:
    def __init__(self, x=0.0, y=0.0):
        self.x = x
        self.y = y

    def __str__(self):
        return "[{}, {}]".format(str(self.x), str(self.y))
```

```python
v1 = Vector(2, 8)
print(v1)
```

```python
v2 = Vector(5, 3)
print(v2)
```

```python
def add(self, v2):
    v3 = Vector()
    v3.x = self.x + v2.x
    v3.y = self.y + v2.y
    return v3

Vector.add = add
```

```python
z = v1.add(v2)
print(z)
```

```python
def mul(self, s):
    return Vector(s * self.x, s * self.y)

Vector.mul = mul
```

```python
a = v1.mul(5)
print(a)
```

```python
def sub(self, v2):
    return self.add(v2.mul(-1))

Vector.sub = sub
```

```python
a_min_v2 = a.sub(v2)
print(a_min_v2)
```

## 1.2. Matrix Representation (Dense)

```python
class Matrix:
    def __init__(self, dims, fill):
        self.rows = dims[0]
        self.cols = dims[1]
        
        # List Comprehension
        self.A = [
            [fill] * self.cols
            for i in range(self.rows)
        ]
```

```python
m = Matrix((3, 4), 3.5)
```

```python
print(m)
```

```python
def __str__(self):
    rows = len(self.A)
    ret = ''

    for i in range(rows):
        cols = len(self.A[i])

        for j in range(cols):
            ret += str(self.A[i][j]) + "\t"
        ret += "\n"

    return ret

Matrix.__str__ = __str__
```

```python
print(m)
```

```python
%time n = Matrix((100, 100), 0.0)
```

```python
#recall that we can get values from our matrix using indices
def get(self, i, j):
    if i < 0 or i > self.rows:
        raise ValueError("Row idex out of range.")
    if j < 0 or j > self.cols:
        raise ValueError("Column index out of range.")

    return self.A[i][j]

Matrix.get = get
```

## 1.2.2. Matrix Representation (Sparse)

```python
class Matrix:
    def __init__(self, dims):
        self.rows = dims[0]
        self.cols = dims[1]
        self.vals = {}
```

```python
def set(self, i, j, val):
    self.vals[(i, j)] = val

Matrix.set = set
```

```python
def get(self, i, j):
    if i < 0 or i > self.rows:
        raise ValueError("Row index out of range.")
    if j < 0 or j > self.cols:
        raise ValueError("Column index out of range.")

    if (i, j) in self.vals:
        return self.vals[(i, j)]

    return 0.0

Matrix.get = get
```

```python
m = Matrix((5, 5))
```

```python
print(m.vals)
```

```python
m.get(1, 1)
```

```python
m.get(10, 5)
```

```python
m.set(1, 2, 15.0)
```

```python
m.get(1, 2)
```
